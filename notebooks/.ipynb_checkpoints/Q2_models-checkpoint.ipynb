# Imports

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
import os
import warnings
from statsmodels.tsa.arima.model import ARIMA
warnings.filterwarnings("ignore")
from xgboost import XGBRegressor
from itertools import product
import tensorflow as tf
from tensorflow import keras
from sklearn.preprocessing import StandardScaler
from statsmodels.stats.diagnostic import het_arch, acorr_ljungbox

OUTPUT_PATH_figures = "/Users/chamika/Desktop/git/machine_learning/Question_2/outputs/figures/"
DATA_PATH_raw    = "/Users/chamika/Desktop/git/machine_learning/Question_2/data/raw/"
DATA_PATH_processed    = "/Users/chamika/Desktop/git/machine_learning/Question_2/data/processed/"
```

# Model Development: Traditional and Deep Learning Models 

```python
# Cell 9: Section 3 setup — time-aware train/val/test split

data = pd.read_csv(
    os.path.join(DATA_PATH_processed, "data_cleaned.csv"),
    index_col=0, parse_dates=True
)

# Calendar-based chronological split (no shuffling — temporal order preserved)
train = data.loc[:'2021-12-31']
val   = data.loc['2022-01-01':'2022-12-31']
test  = data.loc['2023-01-01':]

for name, d in [("Train", train), ("Val", val), ("Test", test)]:
    print(f"{name:5s}: {d.index.min().date()} to {d.index.max().date()}  "
          f"({len(d)} obs, {len(d)/len(data)*100:.1f}%)")
```

```python
# Cell 10: Naive persistence baseline (benchmark all models must beat)

def evaluate(y_true, y_pred, name):
    y_true, y_pred = np.asarray(y_true), np.asarray(y_pred)
    mae  = np.mean(np.abs(y_true - y_pred))
    rmse = np.sqrt(np.mean((y_true - y_pred) ** 2))
    mape = np.mean(np.abs((y_true - y_pred) / y_true)) * 100
    print(f"{name:25s}  MAE: {mae:8.2f}   RMSE: {rmse:8.2f}   MAPE: {mape:.3f}%")
    return {"Model": name, "MAE": mae, "RMSE": rmse, "MAPE": mape}

results = []

# Persistence: predicted close(t+1) = close(t)
naive_pred = test['Close'].values          # today's close
naive_true = test['target_close_t+1'].values
results.append(evaluate(naive_true, naive_pred, "Naive (persistence)"))
```

# 1. ARIMA

```python
# Cell 11: ARIMA order selection — grid search (p, q) by AIC on train set

import warnings
from statsmodels.tsa.arima.model import ARIMA
warnings.filterwarnings("ignore")

log_close_train = np.log(train['Close'])

aic_results = []
for p in range(4):
    for q in range(4):
        try:
            fit = ARIMA(log_close_train, order=(p, 1, q)).fit()
            aic_results.append({"p": p, "q": q, "AIC": fit.aic})
        except Exception:
            continue

aic_df = pd.DataFrame(aic_results).sort_values("AIC").reset_index(drop=True)
print(aic_df.head(8))

best_p, best_q = int(aic_df.loc[0, "p"]), int(aic_df.loc[0, "q"])
print(f"\nSelected order: ARIMA({best_p},1,{best_q}) by minimum AIC")
```

```python
# Cell 12: ARIMA walk-forward one-step-ahead forecasting on test set

# Fit final model on train + val (all data before the test period).
# Fit on raw values (integer index): trading days have irregular calendar
# gaps, so statsmodels can't infer a date frequency for .append()
log_close_trainval = np.log(pd.concat([train, val])['Close']).values
arima_fit = ARIMA(log_close_trainval, order=(best_p, 1, best_q)).fit()

# Walk forward through test set: forecast 1 step, then append the actual
# observation (refit=False keeps parameters fixed — no test leakage into
# estimation, but the model always conditions on the latest real data,
# exactly as it would in live deployment)
log_test_close = np.log(test['Close']).values
arima_preds_log = []

current_fit = arima_fit
for actual_log in log_test_close:
    fc = current_fit.forecast(steps=1)
    arima_preds_log.append(fc[0])
    current_fit = current_fit.append([actual_log], refit=False)

# Convert back from log scale
arima_preds = np.exp(np.array(arima_preds_log))

# Each forecast is for the corresponding test day, made using info up to
# the previous day. Evaluate against the actual close:
arima_true = test['Close'].values
results.append(evaluate(arima_true, arima_preds, f"ARIMA({best_p},1,{best_q})"))
```

# 2. XGBoost

```python
# Cell 13: XGBoost — predict next-day LOG RETURN, tune on validation set

from xgboost import XGBRegressor
from itertools import product

# Target: next-day log return (reconstructable to price; avoids the tree
# extrapolation problem — test prices exceed the training range)
FEATURES = [c for c in data.columns if c not in ['target_close_t+1']]

def make_xy(d):
    X = d[FEATURES]
    y = np.log(d['target_close_t+1'] / d['Close'])   # next-day log return
    return X, y

X_train, y_train = make_xy(train)
X_val,   y_val   = make_xy(val)
X_test,  y_test  = make_xy(test)

# Small grid over the parameters that matter most for this data size
param_grid = {
    'n_estimators':  [200, 500],
    'max_depth':     [3, 5],
    'learning_rate': [0.01, 0.05],
    'subsample':     [0.8],
    'colsample_bytree': [0.8],
}

tuning_results = []
keys = list(param_grid.keys())
for combo in product(*param_grid.values()):
    params = dict(zip(keys, combo))
    model = XGBRegressor(**params, random_state=42, n_jobs=-1)
    model.fit(X_train, y_train, verbose=False)

    # Validation: reconstruct price from predicted return, score in price space
    val_ret_pred = model.predict(X_val)
    val_price_pred = val['Close'].values * np.exp(val_ret_pred)
    val_rmse = np.sqrt(np.mean((val['target_close_t+1'].values - val_price_pred) ** 2))
    tuning_results.append({**params, 'val_RMSE': val_rmse})

tune_df = pd.DataFrame(tuning_results).sort_values('val_RMSE').reset_index(drop=True)
print(tune_df.head(8))
best_params = tune_df.loc[0, keys].to_dict()
best_params = {k: (int(v) if k in ['n_estimators', 'max_depth'] else float(v))
               for k, v in best_params.items()}
print(f"\nBest params: {best_params}")
```

```python
# Cell 14: XGBoost — final fit on train+val, evaluate once on test

trainval = pd.concat([train, val])
X_trainval, y_trainval = make_xy(trainval)

xgb_final = XGBRegressor(**best_params, random_state=42, n_jobs=-1)
xgb_final.fit(X_trainval, y_trainval, verbose=False)

# Predict next-day return, reconstruct next-day price
test_ret_pred = xgb_final.predict(X_test)
xgb_price_pred = test['Close'].values * np.exp(test_ret_pred)

xgb_true = test['target_close_t+1'].values
results.append(evaluate(xgb_true, xgb_price_pred, "XGBoost (return target)"))
```

```python
# Cell 14b: XGBoost feature importances — testing the calendar-feature claim

importances = pd.Series(
    xgb_final.feature_importances_, index=FEATURES
).sort_values(ascending=False)

fig, ax = plt.subplots(figsize=(10, 6))
importances.head(15).iloc[::-1].plot.barh(ax=ax, color='steelblue')
ax.set_title('XGBoost feature importances — top 15 (gain-based)')
ax.set_xlabel('Importance')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_PATH_figures, "q2_xgb_importances.png"), dpi=150)
plt.show()

# Aggregate importance of the calendar indicators specifically
calendar = ['day_of_week', 'month', 'quarter']
print(f"Calendar features combined importance: {importances[calendar].sum():.4f} "
      f"({importances[calendar].sum()*100:.2f}% of total)")
print(importances[calendar].round(4).to_string())
```

# 3. LSTM

```python
# Cell 15: LSTM — data preparation (sliding windows of past log returns)

import tensorflow as tf
from tensorflow import keras
from sklearn.preprocessing import StandardScaler

tf.random.set_seed(42)
np.random.seed(42)

WINDOW = 21  # one trading month of history per prediction

# Continuous return series across the full period (windows must be able to
# span split boundaries: the first val prediction needs late-train history)
log_ret_full = np.log(data['Close'] / data['Close'].shift(1)).dropna()

# Scale using TRAIN statistics only (no leakage)
scaler = StandardScaler()
scaler.fit(log_ret_full.loc[:'2021-12-31'].values.reshape(-1, 1))
ret_scaled = pd.Series(
    scaler.transform(log_ret_full.values.reshape(-1, 1)).flatten(),
    index=log_ret_full.index
)

def make_windows(series, start, end, window=WINDOW):
    """X = window of past scaled returns, y = next scaled return.
    Prediction dates restricted to [start, end]; history may reach back
    before `start` (that's fine — it's past information, not leakage)."""
    X, y, dates = [], [], []
    values, idx = series.values, series.index
    for i in range(window, len(series)):
        d = idx[i]
        if start <= d <= end:
            X.append(values[i - window:i])
            y.append(values[i])
            dates.append(d)
    return np.array(X)[..., np.newaxis], np.array(y), pd.DatetimeIndex(dates)

X_tr, y_tr, dates_tr = make_windows(ret_scaled, data.index.min(), pd.Timestamp('2021-12-31'))
X_v,  y_v,  dates_v  = make_windows(ret_scaled, pd.Timestamp('2022-01-01'), pd.Timestamp('2022-12-31'))
X_te, y_te, dates_te = make_windows(ret_scaled, pd.Timestamp('2023-01-01'), data.index.max())

print(f"Train windows: {X_tr.shape}   Val: {X_v.shape}   Test: {X_te.shape}")
```

```python
# Cell 16: LSTM — build, train with early stopping, evaluate on test

model = keras.Sequential([
    keras.layers.Input(shape=(WINDOW, 1)),
    keras.layers.LSTM(32),
    keras.layers.Dropout(0.2),
    keras.layers.Dense(1)
])
model.compile(optimizer=keras.optimizers.Adam(0.001), loss='mse')

early_stop = keras.callbacks.EarlyStopping(
    monitor='val_loss', patience=10, restore_best_weights=True
)

history = model.fit(
    X_tr, y_tr,
    validation_data=(X_v, y_v),
    epochs=100, batch_size=32,
    callbacks=[early_stop], verbose=0
)
print(f"Stopped after {len(history.history['loss'])} epochs "
      f"(best val_loss: {min(history.history['val_loss']):.4f})")

# Predict scaled returns -> unscale -> reconstruct price
test_ret_scaled_pred = model.predict(X_te, verbose=0).flatten()
test_ret_pred_lstm = scaler.inverse_transform(
    test_ret_scaled_pred.reshape(-1, 1)
).flatten()

# Each prediction at date d is the return FROM the previous close TO d's close
prev_close = data['Close'].shift(1).loc[dates_te].values
lstm_price_pred = prev_close * np.exp(test_ret_pred_lstm)
lstm_true = data['Close'].loc[dates_te].values

results.append(evaluate(lstm_true, lstm_price_pred, "LSTM (return target)"))
```

# Volatility Modeling using GARCH 

```python
# Cell 18: Formal test for ARCH effects (conditional heteroscedasticity)

from statsmodels.stats.diagnostic import het_arch, acorr_ljungbox

# Work in percentage returns (x100): standard for the arch library,
# avoids numerical convergence issues with tiny raw-return variances
returns = 100 * np.log(data['Close'] / data['Close'].shift(1)).dropna()

# Visual evidence: autocorrelation of returns vs squared returns
from statsmodels.graphics.tsaplots import plot_acf
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
plot_acf(returns, lags=20, ax=axes[0], title='ACF: log returns')
plot_acf(returns**2, lags=20, ax=axes[1], title='ACF: squared log returns')
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_PATH_figures, "q2_acf_arch_effects.png"), dpi=150)
plt.show()

# Engle's ARCH-LM test (H0: no ARCH effects / constant conditional variance)
lm_stat, lm_p, f_stat, f_p = het_arch(returns, nlags=10)
print(f"ARCH-LM test (10 lags):  LM stat = {lm_stat:.1f}   p-value = {lm_p:.2e}")

# Ljung-Box on squared returns (H0: no autocorrelation in squared returns)
lb = acorr_ljungbox(returns**2, lags=[10], return_df=True)
print(f"Ljung-Box on r² (10 lags): stat = {lb['lb_stat'].iloc[0]:.1f}   "
      f"p-value = {lb['lb_pvalue'].iloc[0]:.2e}")
```

```python
# Cell 19: Fit GARCH(1,1) — parameters estimated on pre-test data only

from arch import arch_model

# Constant mean (Section 2.3 showed the mean is unpredictable);
# Student-t innovations (excess kurtosis of 15.76 rules out normal errors)
am = arch_model(returns, mean='Constant', vol='GARCH', p=1, q=1, dist='t')

# last_obs: parameters estimated using data up to end-2022 only —
# the test period never enters estimation (same discipline as Section 2.3)
garch_res = am.fit(last_obs='2023-01-01', disp='off')
print(garch_res.summary())

alpha = garch_res.params['alpha[1]']
beta  = garch_res.params['beta[1]']
print(f"\nPersistence (alpha + beta): {alpha + beta:.4f}")
print(f"Half-life of volatility shock: {np.log(0.5)/np.log(alpha+beta):.1f} days")
```

```python
# Cell 19b: Robustness — is the boundary persistence driven by COVID?

am_pre = arch_model(returns.loc[:'2019-12-31'], mean='Constant',
                    vol='GARCH', p=1, q=1, dist='t')
res_pre = am_pre.fit(disp='off')
a, b = res_pre.params['alpha[1]'], res_pre.params['beta[1]']
print(f"Pre-COVID sample (2015–2019):  alpha = {a:.4f}   beta = {b:.4f}   "
      f"alpha+beta = {a+b:.4f}")
if a + b < 1:
    print(f"Half-life of volatility shock: {np.log(0.5)/np.log(a+b):.1f} days")
```

```python
# Cell 20: One-day-ahead volatility forecasts vs realised volatility (test period)

# Forecast row indexed t contains the variance forecast FOR t+1, made at t
# (fixed parameters from Cell 19 — walk-forward filtering, no re-estimation)
fc = garch_res.forecast(horizon=1, start='2023-01-01', reindex=False)
garch_vol_fc = np.sqrt(fc.variance['h.1'])          # conditional st.dev, % units

# Align: shift forecast one day forward so each value sits on the day it predicts
garch_vol_fc = garch_vol_fc.shift(1).dropna()

# Realised proxies on the same dates:
abs_ret = returns.loc[garch_vol_fc.index].abs()               # noisy daily proxy
realised_21 = returns.rolling(21).std().loc[garch_vol_fc.index]  # smooth proxy

# Accuracy vs daily proxy
mae_garch = np.mean(np.abs(garch_vol_fc - abs_ret))
naive_vol = returns.std()  # constant (unconditional) volatility benchmark
mae_const = np.mean(np.abs(naive_vol - abs_ret))
print(f"MAE vs |return|:  GARCH(1,1) = {mae_garch:.4f}   "
      f"constant-volatility benchmark = {mae_const:.4f}")
corr = np.corrcoef(garch_vol_fc, realised_21)[0, 1]
print(f"Correlation with 21-day realised volatility: {corr:.3f}")

# Plot
fig, ax = plt.subplots(figsize=(13, 5))
ax.plot(abs_ret.index, abs_ret, color='lightgray', lw=0.8, label='|daily return| (noisy proxy)')
ax.plot(realised_21.index, realised_21, color='black', lw=1.5, label='21-day realised volatility')
ax.plot(garch_vol_fc.index, garch_vol_fc, color='crimson', lw=1.3,
        label='GARCH(1,1) 1-day-ahead forecast')
ax.set_title('GARCH(1,1) conditional volatility forecasts vs realised volatility — test period')
ax.set_ylabel('Daily volatility (%)')
ax.legend()
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_PATH_figures, "q2_garch_forecast.png"), dpi=150)
plt.show()
```

# Model Comparison, Insights, and Critical Reflection

```python
# Cell 21: Synthesis — do point-forecast errors cluster where GARCH predicts?

# Absolute error of the naive forecast, in % of price (comparable across time)
naive_abs_err_pct = 100 * np.abs(
    test['target_close_t+1'].values - test['Close'].values
) / test['Close'].values
err_series = pd.Series(naive_abs_err_pct, index=test.index)

# 21-day rolling MAE of the point forecast vs GARCH conditional volatility
roll_err = err_series.rolling(21).mean()

fig, ax1 = plt.subplots(figsize=(13, 5))
ax1.plot(roll_err.index, roll_err, color='steelblue', lw=1.6,
         label='21-day rolling MAE of persistence forecast (% of price)')
ax1.set_ylabel('Rolling forecast MAE (%)', color='steelblue')
ax1.tick_params(axis='y', labelcolor='steelblue')

ax2 = ax1.twinx()
ax2.plot(garch_vol_fc.index, garch_vol_fc, color='crimson', lw=1.3, alpha=0.8,
         label='GARCH(1,1) forecast volatility (%)')
ax2.set_ylabel('GARCH conditional volatility (%)', color='crimson')
ax2.tick_params(axis='y', labelcolor='crimson')

corr_sync = roll_err.corr(garch_vol_fc.reindex(roll_err.index))
ax1.set_title(f'Point-forecast error vs GARCH volatility forecast — test period '
              f'(correlation: {corr_sync:.3f})')
fig.legend(loc='upper left', bbox_to_anchor=(0.08, 0.95))
plt.tight_layout()
plt.savefig(os.path.join(OUTPUT_PATH_figures, "q2_error_vs_volatility.png"), dpi=150)
plt.show()
print(f"Correlation (rolling point-forecast MAE vs GARCH volatility): {corr_sync:.3f}")
```
